# Exploratory Data Analysis (EDA) Dasar
## Dataset Supermarket Sales Cleaned

### Tujuan Analisis
Analisis ini bertujuan untuk melakukan Exploratory Data Analysis (EDA) dasar pada dataset Supermarket Sales Cleaned. Analisis dilakukan untuk memahami karakteristik dataset menggunakan statistik deskriptif dan eksplorasi data.

Analisis yang dilakukan meliputi:
- Melihat struktur dan informasi dataset
- Menggunakan `describe()`
- Menggunakan `value_counts()`
- Menggunakan `unique()`
- Menggunakan `nunique()`
- Menghitung mean, median, dan modus
- Menghitung minimum dan maksimum
- Menghitung standar deviasi
- Menghitung kuartil


## 1. Import Library


```python
import pandas as pd
import numpy as np
```

## 2. Membaca Dataset


```python
df = pd.read_csv("Supermarket Sales Cleaned.csv")
df.head()
```

## 3. Ukuran Dataset


```python
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
```

## 4. Informasi Dataset
Tahap ini digunakan untuk melihat nama kolom, tipe data, dan jumlah data pada setiap kolom.


```python
df.info()
```

## 5. Nama Kolom


```python
df.columns
```

## 6. Statistik Deskriptif dengan describe()
Fungsi `describe()` digunakan untuk melihat ringkasan statistik data numerik.


```python
df.describe()
```

## 7. Analisis Kategori dengan value_counts()
Fungsi `value_counts()` digunakan untuk menghitung jumlah kemunculan setiap kategori.


```python
print("Jumlah transaksi berdasarkan Branch:")
print(df["Branch"].value_counts())

print("\nJumlah transaksi berdasarkan Product line:")
print(df["Product line"].value_counts())

print("\nJumlah transaksi berdasarkan Payment:")
print(df["Payment"].value_counts())

print("\nJumlah transaksi berdasarkan Gender:")
print(df["Gender"].value_counts())
```

## 8. Nilai Unik dengan unique()


```python
print("Branch:", df["Branch"].unique())
print("Payment:", df["Payment"].unique())
print("Gender:", df["Gender"].unique())
print("Customer type:", df["Customer type"].unique())
```

## 9. Jumlah Nilai Unik dengan nunique()


```python
df.nunique()
```

## 10. Memilih Kolom Numerik
Kolom berikut dipilih untuk analisis statistik dasar.


```python
numeric_columns = [
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Total",
    "cogs",
    "gross income",
    "Rating"
]

df[numeric_columns].head()
```

## 11. Mean (Rata-rata)


```python
df[numeric_columns].mean()
```

## 12. Median
Median adalah nilai tengah setelah data diurutkan.


```python
df[numeric_columns].median()
```

## 13. Modus
Modus adalah nilai yang paling sering muncul.


```python
df[numeric_columns].mode()
```

## 14. Nilai Minimum


```python
df[numeric_columns].min()
```

## 15. Nilai Maksimum


```python
df[numeric_columns].max()
```

## 16. Standar Deviasi
Standar deviasi digunakan untuk melihat tingkat penyebaran data terhadap rata-ratanya.


```python
df[numeric_columns].std()
```

## 17. Kuartil
- Q1 = 25%
- Q2 = 50% atau median
- Q3 = 75%


```python
df[numeric_columns].quantile([0.25, 0.50, 0.75])
```

## 18. Membuat Statistical Summary


```python
statistical_summary = pd.DataFrame({
    "Count": df[numeric_columns].count(),
    "Mean": df[numeric_columns].mean(),
    "Median": df[numeric_columns].median(),
    "Mode": df[numeric_columns].mode().iloc[0],
    "Minimum": df[numeric_columns].min(),
    "Maximum": df[numeric_columns].max(),
    "Standard Deviation": df[numeric_columns].std(),
    "Q1 (25%)": df[numeric_columns].quantile(0.25),
    "Q2 (50%)": df[numeric_columns].quantile(0.50),
    "Q3 (75%)": df[numeric_columns].quantile(0.75)
})

statistical_summary
```

## 19. Export Statistical Summary ke Excel


```python
statistical_summary.to_excel(
    "Statistical Summary.xlsx",
    sheet_name="Statistical Summary"
)

print("File Statistical Summary.xlsx berhasil dibuat!")
```

Berdasarkan Exploratory Data Analysis (EDA) dasar pada dataset Supermarket Sales Cleaned, dataset terdiri dari data transaksi supermarket dengan informasi pelanggan, produk, harga, jumlah pembelian, metode pembayaran, pendapatan, dan rating.

Analisis menggunakan `describe()` memberikan gambaran statistik dasar dari data numerik. Fungsi `value_counts()`, `unique()`, dan `nunique()` membantu memahami kategori dan variasi nilai pada dataset.

Kolom numerik dianalisis menggunakan mean, median, modus, minimum, maksimum, standar deviasi, serta kuartil. Hasil analisis statistik ini dapat digunakan sebagai dasar untuk melakukan visualisasi dan analisis lebih lanjut pada tahap berikutnya.
